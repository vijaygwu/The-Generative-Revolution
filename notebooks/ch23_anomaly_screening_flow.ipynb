This notebook runs the synthetic anomaly-screening workflow, then compares the result against the saved expected artifact for the companion repo.

```python
import json
import math
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from examples.anomaly_screening_flow import run_demo

expected_path = ROOT / "artifacts" / "expected" / "anomaly_screening_flow.json"
expected = json.loads(expected_path.read_text())
result = run_demo(seed=0)
result
```

```python
for key, expected_value in expected.items():
    value = result[key]
    if isinstance(expected_value, float):
        assert math.isclose(value, expected_value, rel_tol=1e-6, abs_tol=1e-6), (key, value, expected_value)
    else:
        assert value == expected_value, (key, value, expected_value)

print("Notebook output matches the saved expected artifact.")
```
