This notebook runs the synthetic multimodal workflow, then compares the result against the saved expected artifact for the companion repo.

```python
import json
import math
import sys
from pathlib import Path

def find_companion_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        for root in (candidate, candidate / "code" / "The-Generative-Revolution", candidate / "The-Generative-Revolution"):
            if (root / "src").is_dir() and (root / "examples").is_dir():
                return root
    raise RuntimeError(
        "Could not locate the Book 3 companion repo root. "
        "Clone or open The-Generative-Revolution, then run this notebook "
        "from the repo root or its notebooks/ directory."
    )

ROOT = find_companion_root(Path.cwd().resolve())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from examples.multimodal_creative_assistant import run_demo

expected_path = ROOT / "artifacts" / "expected" / "multimodal_creative_assistant.json"
expected = json.loads(expected_path.read_text())
result = run_demo(seed=0)
result
```

```python
for key, expected_value in expected.items():
    value = result[key]
    if isinstance(expected_value, float):
        assert math.isclose(value, expected_value, rel_tol=1e-6, abs_tol=1e-6), (key, value, expected_value)
    else:
        assert value == expected_value, (key, value, expected_value)

print("Notebook output matches the saved expected artifact.")
```
