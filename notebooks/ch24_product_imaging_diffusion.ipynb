This notebook runs the synthetic product-imaging workflow, then compares the result against the saved expected artifact for the companion repo.

```python
import json
import math
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from examples.product_imaging_diffusion import run_demo

expected_path = ROOT / "artifacts" / "expected" / "product_imaging_diffusion.json"
expected = json.loads(expected_path.read_text())
result = run_demo(seed=0)
result
```

```python
def compare_dict(actual, expected, prefix=""):
    for key, expected_value in expected.items():
        full_key = f"{prefix}{key}"
        value = actual[key]
        if isinstance(expected_value, dict):
            compare_dict(value, expected_value, prefix=f"{full_key}.")
        elif isinstance(expected_value, float):
            assert math.isclose(value, expected_value, rel_tol=1e-6, abs_tol=1e-6), (full_key, value, expected_value)
        else:
            assert value == expected_value, (full_key, value, expected_value)

compare_dict(result, expected)
print("Notebook output matches the saved expected artifact.")
```
